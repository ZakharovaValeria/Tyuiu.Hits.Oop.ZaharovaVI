Создать базовый класс BankAccount в C#, который будет представлять
информацию об учетных записях в банке. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

Требования к базовому классу BankAccount:
• Атрибуты: Номер счета (AccountNumber), Баланс (Balance), Тип счета
(AccountType).
• Методы:
    o GetInfo(): метод для получения информации о счете в виде строки.
    o Deposit(decimal amount): метод для внесения денег на счет.
    o Withdraw(decimal amount): метод для снятия денег со счета.

Требования к производным классам:
1. Сберегательный счет (SavingsAccount): Должен содержать дополнительные
атрибуты, такие как Процентная ставка (InterestRate). Метод Deposit() должен
быть переопределен для добавления процентов к сумме вклада при
внесении денег на счет.
2. Текущий счет (CheckingAccount): Должен содержать дополнительные
атрибуты, такие как Лимит овердрафта (OverdraftLimit).
Метод Withdraw() должен быть переопределен для проверки и применения
лимита овердрафта при снятии денег со счета.
3. Инвестиционный счет (InvestmentAccount) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Список активов
(AssetsList). Метод GetInfo() должен быть переопределен для включения
информации о списках активов в описании счета.

#### Дополнительное задание

```
using System;
using System.Collections.Generic;
using System.Linq;

public interface IInsurable
{
    void Insure();
}

public interface ITransferable
{
    void Transfer(BankAccount target, decimal amount);
}

public interface IAuditable
{
    void Audit();
}

public delegate void BankTransactionHandler(string accountNumber, string message);

public class BankAccount : IAuditable
{
    private string _accountNumber;
    private decimal _balance;
    private string _accountType;
    private DateTime _openDate;
    private bool _isBlocked;
    private string _bankName;
    private string _currency;

    private string _ownerName;
    private string _phoneNumber;
    private List<string> _transactionHistory;

    public event BankTransactionHandler TransactionOccurred;

    public string AccountNumber { get { return _accountNumber; } set { _accountNumber = value; } }
    public decimal Balance { get { return _balance; } set { _balance = value; } }
    public string AccountType { get { return _accountType; } set { _accountType = value; } }
    public DateTime OpenDate { get { return _openDate; } set { _openDate = value; } }
    public bool IsBlocked { get { return _isBlocked; } set { _isBlocked = value; } }
    public string BankName { get { return _bankName; } set { _bankName = value; } }
    public string Currency { get { return _currency; } set { _currency = value; } }

    public string OwnerName { get { return _ownerName; } set { _ownerName = value; } }
    public string PhoneNumber { get { return _phoneNumber; } set { _phoneNumber = value; } }
    public List<string> TransactionHistory { get { return _transactionHistory; } set { _transactionHistory = value; } }

    public BankAccount(string accountNumber, decimal balance, string accountType)
    {
        AccountNumber = accountNumber;
        Balance = balance;
        AccountType = accountType;
        OpenDate = DateTime.Now;
        IsBlocked = false;
        BankName = "Главный Банк";
        Currency = "RUB";
        OwnerName = "Не указан";
        PhoneNumber = "Не указан";
        TransactionHistory = new List<string>();
    }

    protected virtual void OnTransactionOccurred(string accountNumber, string message)
    {
        TransactionOccurred?.Invoke(accountNumber, message);
    }

    public virtual string GetInfo()
    {
        return $"Счет №{AccountNumber} | Владелец: {OwnerName} | Тип: {AccountType} | Баланс: {Balance} {Currency} | Открыт: {OpenDate:dd.MM.yyyy}";
    }

    public virtual void Deposit(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine($"Ошибка: сумма пополнения должна быть больше нуля. Получено: {amount:C}");
            return;
        }

        Balance += amount;
        TransactionHistory.Add($"Пополнение на {amount:C}. Баланс: {Balance:C}");
        OnTransactionOccurred(AccountNumber, $"Внесено {amount:C}");
        Console.WriteLine($"Внесено {amount:C}. Новый баланс: {Balance:C}");
    }

    public virtual void Withdraw(decimal amount)
    {
        if (amount > Balance)
        {
            Console.WriteLine($"Ошибка: недостаточно средств. Доступно: {Balance}");
        }
        else
        {
            Balance -= amount;
            TransactionHistory.Add($"Снятие {amount:C}. Баланс: {Balance:C}");
            OnTransactionOccurred(AccountNumber, $"Снято {amount:C}");
            Console.WriteLine($"Снято {amount}. Новый баланс: {Balance}");
        }
    }

    public void PrintHistory()
    {
        Console.WriteLine($"\nИстория операций по счету {AccountNumber}");
        if (TransactionHistory.Count == 0)
        {
            Console.WriteLine("Операций не было.");
        }
        foreach (var record in TransactionHistory)
        {
            Console.WriteLine(record);
        }
    }

    public void Block()
    {
        IsBlocked = true;
        Console.WriteLine($"Счет {AccountNumber} заблокирован.");
    }

    public void Unblock()
    {
        IsBlocked = false;
        Console.WriteLine($"Счет {AccountNumber} разблокирован.");
    }

    void IAuditable.Audit()
    {
        Console.WriteLine($"[Аудит] Проверка счета {AccountNumber}. Баланс в норме: {Balance} {Currency}");
    }

    public void PrintWelcomeMessage()
    {
        Console.WriteLine($"Добро пожаловать в {BankName}!");
    }

    public void ChangeCurrency(string newCurrency)
    {
        Currency = newCurrency;
        Console.WriteLine($"Валюта счета {AccountNumber} изменена на {Currency}");
    }
}

public class SavingsAccount : BankAccount
{
    private decimal _interestRate;
    private decimal _minBalance;
    private int _monthsOpened;
    private int _payoutDay;

    private string _goalName;
    private decimal _targetAmount;

    public decimal InterestRate { get { return _interestRate; } set { _interestRate = value; } }
    public decimal MinBalance { get { return _minBalance; } set { _minBalance = value; } }
    public int MonthsOpened { get { return _monthsOpened; } set { _monthsOpened = value; } }
    public int PayoutDay { get { return _payoutDay; } set { _payoutDay = value; } }
    public string GoalName { get { return _goalName; } set { _goalName = value; } }
    public decimal TargetAmount { get { return _targetAmount; } set { _targetAmount = value; } }

    public SavingsAccount(string accountNumber, decimal balance, decimal interestRate)
        : base(accountNumber, balance, "Сберегательный")
    {
        InterestRate = interestRate;
        MinBalance = 100m;
        MonthsOpened = 0;
        PayoutDay = 1;
        GoalName = "Не задана";
        TargetAmount = 0m;
    }

    public override void Deposit(decimal amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine($"[Сберегательный] Ошибка: сумма должна быть больше нуля. Получено: {amount:C}");
            return;
        }

        decimal interest = amount * InterestRate;
        decimal totalAdded = amount + interest;

        Balance += totalAdded;
        TransactionHistory.Add($"Пополнение {amount:C} + проценты {interest:C}. Баланс: {Balance:C}");
        OnTransactionOccurred(AccountNumber, $"Внесено {amount:C} с процентами");

        Console.WriteLine($"[Сберегательный] Внесено {amount:C} + проценты {interest:C} ({InterestRate:P}).");
        Console.WriteLine($"Новый баланс: {Balance:C}");
    }

    public void SetGoal(string name, decimal target)
    {
        GoalName = name;
        TargetAmount = target;
        Console.WriteLine($"Цель счета {AccountNumber}: {GoalName} (Сумма: {TargetAmount:C})");
    }

    public void CheckGoalProgress()
    {
        if (TargetAmount == 0)
        {
            Console.WriteLine($"Для счета {AccountNumber} цель не установлена.");
            return;
        }
        decimal progress = (Balance / TargetAmount) * 100;
        Console.WriteLine($"Прогресс цели '{GoalName}': {progress:F1}% ({Balance:C} из {TargetAmount:C})");
    }

    public decimal CalculateYearlyInterest()
    {
        return Balance * InterestRate * 12;
    }

    public void AddMonth()
    {
        MonthsOpened++;
        Console.WriteLine($"Счет {AccountNumber}: срок вклада — {MonthsOpened} мес.");
    }

    public bool HasMinimumBalance()
    {
        return Balance >= MinBalance;
    }

    public void SetPayoutDay(int day)
    {
        PayoutDay = day;
        Console.WriteLine($"День выплаты процентов установлен на {PayoutDay} число.");
    }

    public void CalculateMonthlyInterest()
    {
        decimal monthly = Balance * (InterestRate / 12);
        Console.WriteLine($"Ожидаемые проценты за месяц: {monthly:C}");
    }
}

public class CheckingAccount : BankAccount
{
    private decimal _overdraftLimit;
    private int _transactionsCount;
    private decimal _monthlyFee;
    private bool _hasCreditLine;
    private decimal _dailyWithdrawalLimit;

    private string _linkedCardNumber;

    public decimal OverdraftLimit { get { return _overdraftLimit; } set { _overdraftLimit = value; } }
    public int TransactionsCount { get { return _transactionsCount; } set { _transactionsCount = value; } }
    public decimal MonthlyFee { get { return _monthlyFee; } set { _monthlyFee = value; } }
    public bool HasCreditLine { get { return _hasCreditLine; } set { _hasCreditLine = value; } }
    public decimal DailyWithdrawalLimit { get { return _dailyWithdrawalLimit; } set { _dailyWithdrawalLimit = value; } }
    public string LinkedCardNumber { get { return _linkedCardNumber; } set { _linkedCardNumber = value; } }

    public CheckingAccount(string accountNumber, decimal balance, decimal overdraftLimit)
        : base(accountNumber, balance, "Текущий")
    {
        OverdraftLimit = overdraftLimit;
        TransactionsCount = 0;
        MonthlyFee = 50m;
        HasCreditLine = false;
        DailyWithdrawalLimit = 10000m;
        LinkedCardNumber = "Не привязана";
    }

    public override void Withdraw(decimal amount)
    {
        if (amount > Balance + OverdraftLimit)
        {
            Console.WriteLine($"[Текущий] Ошибка: превышен лимит овердрафта. Доступно: {Balance + OverdraftLimit}");
        }
        else
        {
            Balance -= amount;
            TransactionsCount++;
            TransactionHistory.Add($"Снятие {amount:C} (овердрафт). Баланс: {Balance:C}");
            OnTransactionOccurred(AccountNumber, $"Снято {amount:C} (Текущий счет)");
            Console.WriteLine($"[Текущий] Снято {amount} (с учетом овердрафта). Новый баланс: {Balance}");
        }
    }

    public void LinkCard(string cardNumber)
    {
        LinkedCardNumber = cardNumber;
        Console.WriteLine($"К счету {AccountNumber} привязана карта: {LinkedCardNumber}");
    }

    public void ShowTransactionCount()
    {
        Console.WriteLine($"Счет {AccountNumber}: выполнено операций — {TransactionsCount}");
    }

    public void ResetTransactionCount()
    {
        TransactionsCount = 0;
        Console.WriteLine($"Счет {AccountNumber}: счетчик операций сброшен.");
    }

    public void ChargeMonthlyFee()
    {
        Balance -= MonthlyFee;
        Console.WriteLine($"Счет {AccountNumber}: списана ежемесячная плата {MonthlyFee:C}. Баланс: {Balance}");
    }

    public void SetDailyWithdrawalLimit(decimal limit)
    {
        DailyWithdrawalLimit = limit;
        Console.WriteLine($"Дневной лимит снятия установлен: {DailyWithdrawalLimit:C}");
    }

    public void CheckDailyLimit(decimal amount)
    {
        if (amount > DailyWithdrawalLimit)
        {
            Console.WriteLine($"Внимание: сумма {amount:C} превышает дневной лимит {DailyWithdrawalLimit:C}!");
        }
        else
        {
            Console.WriteLine($"Сумма {amount:C} в пределах дневного лимита.");
        }
    }
}

public class InvestmentAccount : BankAccount
{
    private List<string> _assetsList;
    private decimal _riskLevel;
    private decimal _dividendRate;
    private DateTime _lastRebalance;
    private string _advisorName;

    private string _brokerName;

    public List<string> AssetsList { get { return _assetsList; } set { _assetsList = value; } }
    public decimal RiskLevel { get { return _riskLevel; } set { _riskLevel = value; } }
    public decimal DividendRate { get { return _dividendRate; } set { _dividendRate = value; } }
    public DateTime LastRebalance { get { return _lastRebalance; } set { _lastRebalance = value; } }
    public string AdvisorName { get { return _advisorName; } set { _advisorName = value; } }
    public string BrokerName { get { return _brokerName; } set { _brokerName = value; } }

    public InvestmentAccount(string accountNumber, decimal balance, List<string> assets)
        : base(accountNumber, balance, "Инвестиционный")
    {
        AssetsList = assets;
        RiskLevel = 0.5m;
        DividendRate = 0.03m;
        LastRebalance = DateTime.Now;
        AdvisorName = "Не назначен";
        BrokerName = "Брокер по умолчанию";
    }

    public override string GetInfo()
    {
        string assets = AssetsList.Count > 0 ? string.Join(", ", AssetsList) : "нет активов";
        return $"{base.GetInfo()} | Брокер: {BrokerName} | Активы: {assets} | Риск: {RiskLevel:P}";
    }

    public void AssignBroker(string name)
    {
        BrokerName = name;
        Console.WriteLine($"Счет {AccountNumber} переведен на обслуживание к брокеру: {BrokerName}");
    }

    public void AddAsset(string asset)
    {
        AssetsList.Add(asset);
        Console.WriteLine($"Актив \"{asset}\" добавлен в портфель счета {AccountNumber}");
    }

    public void RemoveAsset(string asset)
    {
        if (AssetsList.Remove(asset))
        {
            Console.WriteLine($"Актив \"{asset}\" удален из портфеля счета {AccountNumber}");
        }
        else
        {
            Console.WriteLine($"Актив \"{asset}\" не найден в портфеле.");
        }
    }

    public decimal CalculateExpectedDividends()
    {
        return Balance * DividendRate;
    }

    public void Rebalance()
    {
        LastRebalance = DateTime.Now;
        Console.WriteLine($"Портфель счета {AccountNumber} перебалансирован {LastRebalance:dd.MM.yyyy HH:mm}");
    }

    public void AssignAdvisor(string name)
    {
        AdvisorName = name;
        Console.WriteLine($"За счётом {AccountNumber} закреплен советник: {AdvisorName}");
    }

    public void ShowAdvisorInfo()
    {
        Console.WriteLine($"Ваш персональный советник по счету {AccountNumber}: {AdvisorName}");
    }
}

public class PremiumInvestmentAccount : InvestmentAccount, IInsurable, ITransferable
{
    private decimal _insuranceAmount;
    private bool _isInsured;
    private string _conciergePhone;

    public decimal InsuranceAmount { get { return _insuranceAmount; } set { _insuranceAmount = value; } }
    public bool IsInsured { get { return _isInsured; } set { _isInsured = value; } }
    public string ConciergePhone { get { return _conciergePhone; } set { _conciergePhone = value; } }

    public PremiumInvestmentAccount(string accountNumber, decimal balance, List<string> assets)
        : base(accountNumber, balance, assets)
    {
        InsuranceAmount = 0m;
        IsInsured = false;
        RiskLevel = 0.2m;
        ConciergePhone = "+7 (800) 555-35-35";
    }

    public void Insure()
    {
        IsInsured = true;
        InsuranceAmount = Balance * 0.1m;
        Console.WriteLine($"[Премиум] Счет {AccountNumber} застрахован на сумму {InsuranceAmount:C}");
    }

    public void Transfer(BankAccount target, decimal amount)
    {
        if (amount > Balance)
        {
            Console.WriteLine($"[Премиум] Ошибка: недостаточно средств для перевода. Доступно: {Balance}");
            return;
        }

        Balance -= amount;
        target.Balance += amount;
        TransactionHistory.Add($"Перевод {amount:C} на счет {target.AccountNumber}");
        OnTransactionOccurred(AccountNumber, $"Перевод {amount:C}");
        Console.WriteLine($"[Премиум] Переведено {amount:C} со счета {AccountNumber} на счет {target.AccountNumber}");
    }

    public override string GetInfo()
    {
        string insured = IsInsured ? $" | Страховка: {InsuranceAmount:C}" : " | Не застрахован";
        return $"{base.GetInfo()}{insured}";
    }

    public void SetConciergePhone(string phone)
    {
        ConciergePhone = phone;
        Console.WriteLine($"Номер консьерж-сервиса обновлен: {ConciergePhone}");
    }

    public void CallConcierge()
    {
        Console.WriteLine($"Звонок в премиум-консьерж сервис по номеру {ConciergePhone}...");
    }
}

public class BankAuditService
{
    private readonly IAuditable _account;
    public BankAuditService(IAuditable account)
    {
        _account = account;
    }

    public void RunAudit()
    {
        Console.WriteLine("Запуск налогового аудита");
        _account.Audit();
    }
}

Console.WriteLine("Демонстрация работы банковских счетов\n");

SavingsAccount savings = new SavingsAccount("SAV-001", 1000m, 0.05m);
savings.OwnerName = "Иванов И.И.";
savings.PhoneNumber = "+7 (999) 123-45-67";

savings.TransactionOccurred += (accountNum, message) => 
{
    Console.WriteLine($"[Лямбда-событие] Счет {accountNum}: {message}");
};

Console.WriteLine("Создан: " + savings.GetInfo());
savings.SetGoal("Отпуск", 50000m);
savings.Deposit(500m);
savings.SetPayoutDay(10);
savings.CalculateMonthlyInterest();
savings.CheckGoalProgress();
savings.PrintHistory();
Console.WriteLine();

CheckingAccount checking = new CheckingAccount("CHK-002", 500m, 300m);
checking.OwnerName = "Петров П.П.";
Console.WriteLine("Создан: " + checking.GetInfo());
checking.LinkCard("1234 5678 9012 3456");
checking.SetDailyWithdrawalLimit(5000m);
checking.CheckDailyLimit(700m);
checking.PrintHistory();
Console.WriteLine();

InvestmentAccount investment = new InvestmentAccount("INV-003", 5000m, new List<string> { "Акции Apple", "Облигации", "Золото" });
investment.AssignBroker("Финам");
Console.WriteLine("Создан: " + investment.GetInfo());
investment.AssignAdvisor("Иванов И.И.");
investment.ShowAdvisorInfo();
Console.WriteLine();

PremiumInvestmentAccount premium = new PremiumInvestmentAccount("PRM-004", 10000m, new List<string> { "Акции Google", "ETF" });
Console.WriteLine("Создан: " + premium.GetInfo());
premium.CallConcierge();
premium.ChangeCurrency("USD");
premium.PrintWelcomeMessage();
Console.WriteLine();

Console.WriteLine("Явная реализация интерфейса");
IAuditable auditableAccount = checking;
auditableAccount.Audit();
Console.WriteLine();

Console.WriteLine("Управление зависимостями");
BankAuditService auditService = new BankAuditService(savings);
auditService.RunAudit();
```
