```
using System;
using System.Collections.Generic;

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    List<string> GetTeamMembers();
}

public class Employee
{
    private string _name;
    private string _position;
    private double salary;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public double Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name} | Должность: {Position} | Зарплата: {Salary:C}");
    }
}

public class Manager : Employee
{
    private string _department;

    public string Department
    {
        get { return _department; }
        set { _department = value; }
    }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    private string _projectName;
    private List<string> _teamMembers;

    public string ProjectName
    {
        get { return _projectName; }
        set { _projectName = value; }
    }

    public List<string> TeamMembers
    {
        get { return _teamMembers; }
        set { _teamMembers = value; }
    }

    public ProjectManager(string name, string position, double salary, string department, string projectName, List<string> teamMembers)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        TeamMembers = teamMembers;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"Отчет по проекту \"{ProjectName}\" | Руководитель: {Name} | Отдел: {Department} | Команда: {TeamMembers.Count} чел.";
    }

    public List<string> GetTeamMembers()
    {
        return TeamMembers;
    }
}

Console.WriteLine("=== Спринт ревью №2 ===\n");

Employee employee = new Employee("Иванов Иван", "Специалист", 50000);
Console.WriteLine("--- Employee ---");
employee.PrintInfo();
Console.WriteLine();

Manager manager = new Manager("Петров Петр", "Менеджер", 80000, "Продажи");
Console.WriteLine("--- Manager ---");
manager.PrintInfo();
Console.WriteLine();

ProjectManager projectManager = new ProjectManager(
    "Сидоров Сидор",
    "Проектный менеджер",
    120000,
    "Разработка",
    "Внедрение CRM",
    new List<string> { "Анна", "Борис", "Виктор" }
);
Console.WriteLine("--- ProjectManager ---");
projectManager.PrintInfo();
Console.WriteLine();

Console.WriteLine("--- Реализация интерфейсов ---");
Console.WriteLine($"Отчет: {projectManager.GenerateReport()}");
Console.Write("Команда: ");
Console.WriteLine(string.Join(", ", projectManager.GetTeamMembers()));
```
