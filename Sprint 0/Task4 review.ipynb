<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

разработать проект для управления растениями в саду, используя основные концепции объектно-ориентированного программирования в C#. Нужно создать классы, которые будут описывать разные типы растений, используя наследование и полиморфизм.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение [{Name}] цветет.");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"Растение [{Name}] выросло на {growth} см. Новая высота: {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок [{Name}] распускается");
        Console.WriteLine($"  Цвет: {Color}, лепестков: {PetalCount}");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок [{Name}] начинает увядать...");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево [{Name}] покрывается новыми листьями!");
        Console.WriteLine($"  Возраст: {Age} лет, тип листьев: {LeafType}");
    }

    public void Grow(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево [{Name}] выросло на {years} лет. Новый возраст: {Age} лет.");
    }
}

Console.WriteLine("=== Демонстрация работы классов растений ===\n");

Flower rose = new Flower("Роза", 30f, "Красный", 25);
Console.WriteLine($"Создан цветок: {rose.Name}, высота: {rose.Height} см\n");

Tree oak = new Tree("Дуб", 500f, 50, "Лиственные");
Console.WriteLine($"Создано дерево: {oak.Name}, высота: {oak.Height} см\n");

Console.WriteLine("--- Работа с цветком ---");
rose.Grow(5f);
rose.Bloom();
rose.Wither();
Console.WriteLine();

Console.WriteLine("--- Работа с деревом ---");
oak.Grow(2f);
oak.Grow(10);
oak.Bloom();
Console.WriteLine();

Console.WriteLine("--- Полиморфизм: список растений ---");
var garden = new List<Plant> { rose, oak };

foreach (var plant in garden)
{
    Console.WriteLine($"Растение: {plant.Name} (тип: {plant.Type})");
    plant.Bloom();
    Console.WriteLine();
}
```
